```python
import pandas as pd
import matplotlib.pyplot as plt

# Load your dataset
df = pd.read_csv('GA_features.csv')

# Race columns
race_columns = ['registered_A', 'registered_B', 'registered_C', 
                'registered_D', 'registered_E', 'registered_F', 'registered_G']

# Function to count the proportion of interventions per race for each tau
def count_race_proportions(interventions, df):
    race_counts = {race: 0 for race in race_columns}  # Initialize race counts

    # Count interventions per race
    for county_id in interventions:
        for race in race_columns:
            race_counts[race] += df.loc[county_id, race]

    # Convert counts to proportions
    total_interventions = sum(race_counts.values())
    if total_interventions > 0:
        race_proportions = {race.replace('registered_', ''): count / total_interventions for race, count in race_counts.items()}
    else:
        race_proportions = {race.replace('registered_', ''): 0 for race in race_columns}

    return race_proportions

# Compute racial proportions for all counties (Baseline)
total_population = df[race_columns].sum()
baseline_proportions = {race.replace('registered_', ''): total_population[race] / total_population.sum() for race in race_columns}

# Compute racial proportions for each tau value
race_proportion_data = {'Baseline': baseline_proportions}
for tau, counties in interventions_per_tau.items():
    if tau <= 0.45:  # Keep only tau values up to 0.45
        race_proportion_data[tau] = count_race_proportions(counties, df)

# Convert dictionary to DataFrame
race_proportion_df = pd.DataFrame(race_proportion_data).T

# Ensure sorting: Keep 'Baseline' first, then sort tau values numerically
sorted_taus = sorted([tau for tau in race_proportion_df.index if tau != 'Baseline'])
race_proportion_df = race_proportion_df.loc[['Baseline'] + sorted_taus]

# Plotting the results
fig, ax = plt.subplots(figsize=(12, 6), dpi=300)

# Plot as side-by-side bars
bar_width = 0.12  # Adjust for spacing
colors = plt.cm.tab10.colors  # Get colors from tab10 colormap
x = range(len(race_proportion_df.index))

bars = []
for i, race in enumerate(race_proportion_df.columns):
    bars.append(ax.bar(
        [pos + i * bar_width for pos in x], 
        race_proportion_df[race], 
        width=bar_width, 
        color=colors[i % len(colors)], 
        edgecolor='white', 
        alpha=0.8, 
        label=race
    ))

# Annotate the proportion values on the bars
for bar_group in bars:
    for bar in bar_group:
        height = bar.get_height()
        if height > 0:
            ax.text(bar.get_x() + bar.get_width() / 2, height, f'{height:.2f}', 
                    ha='center', va='bottom', fontsize=9, color='black')

# Adding labels, title, and formatting
ax.set_ylabel('Proportion of Interventions', fontsize=12)
ax.set_xlabel('Tau', fontsize=12)
ax.set_title('Proportion of Interventions by Race Across Tau Values', fontsize=14)
ax.legend(title="Race", loc='upper left', fontsize=10)

# Adjust x-ticks for grouped bars
ax.set_xticks([pos + (len(race_proportion_df.columns) / 2) * bar_width for pos in x])
ax.set_xticklabels([str(tau) for tau in race_proportion_df.index], rotation=0, ha='center')

# Adding a grid to make it more readable
ax.grid(axis='y', linestyle='--', linewidth=0.5)

# Adjust layout
plt.tight_layout()
plt.savefig('results/race_proportion_annotated.png', dpi=300)
plt.show()
```
